```python
df['Age'].describe()
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df['Age'], bins=10)

plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Customers')

plt.show()
```

```python
age_bins = [17, 25, 35, 45, 55, 64]

age_labels = [
    '18-25',
    '26-35',
    '36-45',
    '46-55',
    '56-64'
]

df['Age Group'] = pd.cut(
    df['Age'],
    bins=age_bins,
    labels=age_labels
)
```

```python
age_group_counts = df["Age Group"].value_counts().sort_index()

plt.figure(figsize=(8, 5))
age_group_counts.plot(kind="bar")

plt.title("Customer Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
quantity_sales = df.groupby("Product Category")["Quantity"].sum()

plt.figure(figsize=(8, 5))
quantity_sales.plot(kind="bar")

plt.title("Quantity Sold by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
total_sales = df["Total Amount"].sum()
total_quantity = df["Quantity"].sum()
avg_transaction = df["Total Amount"].mean()

print("Total Sales:", total_sales)
print("Total Quantity Sold:", total_quantity)
print("Average Transaction Value:", round(avg_transaction, 2))
```

```python
category_sales = df.groupby("Product Category")["Total Amount"].sum()

best_category = category_sales.idxmax()
best_category_sales = category_sales.max()

print("Best Performing Category:", best_category)
print("Sales:", best_category_sales)
```

```python
lowest_category = category_sales.idxmin()
lowest_category_sales = category_sales.min()

print("Lowest Performing Category:", lowest_category)
print("Sales:", lowest_category_sales)
```

```python
gender_sales = df.groupby("Gender")["Total Amount"].sum()

print(gender_sales)

print("\nHighest spending gender:", gender_sales.idxmax())
print("Sales:", gender_sales.max())
```

```python
monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Total Amount"].sum()

best_month = monthly_sales.idxmax()
best_month_sales = monthly_sales.max()

print("Best Sales Month:", best_month)
print("Sales:", best_month_sales)
```

```python
age_group_counts = df["Age Group"].value_counts()

print("Most active age group:", age_group_counts.idxmax())
print("Transactions:", age_group_counts.max())
```

```python
summary = {
    "Metric": [
        "Total Sales",
        "Total Quantity Sold",
        "Average Transaction Value",
        "Best Product Category",
        "Best Category Sales",
        "Highest Spending Gender",
        "Highest Gender Sales",
        "Best Sales Month",
        "Best Month Sales",
        "Most Active Age Group"
    ],
    "Value": [
        total_sales,
        total_quantity,
        avg_transaction,
        best_category,
        best_category_sales,
        gender_sales.idxmax(),
        gender_sales.max(),
        str(best_month),
        best_month_sales,
        age_group_counts.idxmax()
    ]
}

summary_df = pd.DataFrame(summary)

summary_df
```
